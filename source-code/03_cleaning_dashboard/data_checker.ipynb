```python

#Kiểm tra chất lượng dữ liệu sau khi ghép
import pandas as pd
from pathlib import Path

FILE_PATH = Path("../../dataset/cleaned/products_cleaned.csv")

df = pd.read_csv(FILE_PATH)

print("===== TỔNG QUAN DỮ LIỆU =====")
print("Số dòng:", len(df))
print("Số cột:", len(df.columns))
print("Danh sách cột:")
print(df.columns.tolist())

print("\n===== SỐ SẢN PHẨM THEO NGUỒN =====")
print(df["Nguồn dữ liệu"].value_counts(dropna=False))

print("\n===== GIÁ TRỊ THIẾU =====")
print(df.isna().sum().sort_values(ascending=False))

print("\n===== KIỂM TRA TRÙNG MÃ SẢN PHẨM =====")
print(
    "Số dòng trùng theo nguồn và mã sản phẩm:",
    df.duplicated(
        subset=["Nguồn dữ liệu", "Mã sản phẩm"]
    ).sum()
)

print("\n===== KIỂU DỮ LIỆU =====")
print(df.dtypes)

display(df.head())
```

```python
#Kiểm tra số dòng trùng lặp hoàn toàn
exact_duplicates = df.duplicated().sum()

print("Số dòng trùng lặp hoàn toàn:", exact_duplicates)
```

```python
#Kiểm tra trùng mã sản phẩm trong cùng một nguồn
required_cols = ["Nguồn dữ liệu", "Mã sản phẩm"]
if all(col in df.columns for col in required_cols):
    duplicate_mask = df.duplicated(
        subset=required_cols,
        keep=False
    )

    duplicate_products = df.loc[
        duplicate_mask
    ].sort_values(required_cols)

    print("Số dòng thuộc nhóm trùng mã trong cùng nguồn:",
          len(duplicate_products))

    display(duplicate_products.head(20))
```

```python
# Kiểm tra giá bán, giá gốc và mức giảm giá
price_cols = [
    "Giá bán",
    "Giá gốc",
    "Số tiền giảm giá",
    "Tỷ lệ giảm giá"
]

# Chuyển các cột giá sang dạng số để kiểm tra
for col in price_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Kiểu dữ liệu sau chuyển đổi:")
print(df[price_cols].dtypes)

print("\nThống kê các cột giá:")
display(df[price_cols].describe().T)
```
