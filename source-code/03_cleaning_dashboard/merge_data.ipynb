```python

import pandas as pd
from pathlib import Path

# ==================================================
# 1. KHAI BÁO ĐƯỜNG DẪN
# ==================================================

DATA_DIR = Path("../../dataset")
CLEANED_DIR = DATA_DIR / "cleaned"

TIKI_FILE = CLEANED_DIR / "tiki_cleaned.csv"
HASAKI_FILE = CLEANED_DIR / "hasaki_cleaned.csv"

# Lưu file ghép ngay trong thư mục cleaned
OUTPUT_FILE = CLEANED_DIR / "products_cleaned.csv"


# ==================================================
# 2. KIỂM TRA FILE ĐẦU VÀO
# ==================================================

for file_path in [TIKI_FILE, HASAKI_FILE]:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Không tìm thấy file: {file_path.resolve()}\n"
            "Hãy kiểm tra lại đường dẫn và tên file."
        )

print("Đã tìm thấy cả hai file đầu vào.")


# ==================================================
# 3. ĐỌC DỮ LIỆU
# ==================================================

df_tiki = pd.read_csv(TIKI_FILE)
df_hasaki = pd.read_csv(HASAKI_FILE)

print("\n========== DỮ LIỆU BAN ĐẦU ==========")
print("Tiki:", df_tiki.shape)
print("Hasaki:", df_hasaki.shape)


# ==================================================
# 4. KHAI BÁO 19 CỘT CHUẨN
# ==================================================

standard_columns = [
    "Nguồn dữ liệu",
    "Mã sản phẩm",
    "Tên sản phẩm",
    "Thương hiệu",
    "Giá bán",
    "Giá gốc",
    "Số tiền giảm giá",
    "Tỷ lệ giảm giá",
    "Điểm đánh giá trung bình",
    "Số lượt đánh giá",
    "Số lượng đã bán",
    "Mã danh mục",
    "Tên danh mục",
    "Đường dẫn danh mục",
    "Đường dẫn sản phẩm",
    "Đường dẫn ảnh sản phẩm",
    "Thời gian thu thập"
]


# ==================================================
# 5. KIỂM TRA TÊN CỘT
# ==================================================

def check_columns(df, source_name):
    missing = [
        col for col in standard_columns
        if col not in df.columns
    ]

    extra = [
        col for col in df.columns
        if col not in standard_columns
    ]

    print(f"\n--- {source_name} ---")
    print("Số cột hiện có:", len(df.columns))
    print("Cột thiếu:", missing)
    print("Cột ngoài bộ chuẩn:", extra)

    if missing:
        raise ValueError(
            f"{source_name} thiếu các cột chuẩn: {missing}\n"
            "Hãy chuẩn hóa tên cột trong file nguồn trước khi ghép."
        )


check_columns(df_tiki, "Tiki")
check_columns(df_hasaki, "Hasaki")


# ==================================================
# 6. SẮP XẾP CỘT THEO CÙNG THỨ TỰ
# ==================================================

df_tiki = df_tiki[standard_columns].copy()
df_hasaki = df_hasaki[standard_columns].copy()


# ==================================================
# 7. GHÉP DỌC HAI BỘ DỮ LIỆU
# ==================================================

df_merged = pd.concat(
    [df_tiki, df_hasaki],
    axis=0,
    ignore_index=True
)


# ==================================================
# 8. KIỂM TRA KẾT QUẢ SAU KHI GHÉP
# ==================================================

print("\n========== KẾT QUẢ SAU KHI GHÉP ==========")

print("Tổng số dòng:", len(df_merged))
print("Tổng số cột:", len(df_merged.columns))

print("\nSố sản phẩm theo nguồn:")
print(
    df_merged["Nguồn dữ liệu"]
    .value_counts(dropna=False)
)

print("\nSố giá trị thiếu theo từng cột:")
print(
    df_merged.isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nSố dòng trùng hoàn toàn:", df_merged.duplicated().sum())


# ==================================================
# 9. XUẤT FILE CSV
# ==================================================

df_merged.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

print("\n========== HOÀN TẤT ==========")
print("File đã được lưu tại:")
print(OUTPUT_FILE.resolve())


# ==================================================
# 10. HIỂN THỊ MẪU DỮ LIỆU
# ==================================================

display(df_merged.head(10))
```
